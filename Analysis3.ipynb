```python

import pandas as pd
import numpy as np
import plotly.express as px
from datetime import datetime
from zoneinfo import ZoneInfo

url = "https://raw.githubusercontent.com/vaish-vkb/Google-Play-Store-Analytics/main/Play%20Store%20Data.csv"
df = pd.read_csv(url)

print("Dataset loaded successfully!")
print("Rows and columns:", df.shape)
print("Available columns:")
print(df.columns.tolist())
```

```python

# Clean the required columns

df["App"] = df["App"].astype(str).str.strip()
df["Category"] = df["Category"].astype(str).str.strip()

# Convert Reviews to numeric
df["Reviews"] = pd.to_numeric(df["Reviews"], errors="coerce")

# Clean Installs and convert to numeric
df["Installs"] = (
    df["Installs"].astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)

df["Installs"] = pd.to_numeric(df["Installs"], errors="coerce")

# Convert Last Updated to datetime
df["Last Updated"] = pd.to_datetime(
    df["Last Updated"],
    errors="coerce"
)

# Apply Task 3 filters
filtered_df = df[
    (df["Reviews"] > 500) &
    (~df["App"].str.upper().str.startswith(("X", "Y", "Z"))) &
    (~df["App"].str.upper().str.contains("S", regex=False)) &
    (df["Category"].str.upper().str.startswith(("E", "C", "B")))
].copy()

# Remove rows missing required values
filtered_df = filtered_df.dropna(
    subset=["Installs", "Last Updated", "Category"]
)

print("Rows after filtering:", len(filtered_df))
print("\nCategories included:")
print(sorted(filtered_df["Category"].unique()))
print("\nDate range:")
print(filtered_df["Last Updated"].min(), "to",
      filtered_df["Last Updated"].max())
```

```python

# Translate category names for display on the graph

category_translation = {
    "BEAUTY": "सौंदर्य",
    "BUSINESS": "வணிகம்",
    "DATING": "Partnersuche"
}

filtered_df["Category_Display"] = (
    filtered_df["Category"].map(category_translation)
    .fillna(filtered_df["Category"])
)

# Create month column from Last Updated
filtered_df["Month"] = (
    filtered_df["Last Updated"].dt.to_period("M").dt.to_timestamp()
)

# Calculate total installs by month and category
monthly_data = (
    filtered_df.groupby(["Month", "Category_Display"], as_index=False)
    ["Installs"].sum()
    .sort_values("Month")
)

print("Monthly data created successfully!")
print("Rows:", len(monthly_data))
print("\nSample data:")
display(monthly_data.head(10))
```

```python

# Sort data by category and month
monthly_data = monthly_data.sort_values(
    ["Category_Display", "Month"]
).copy()

# Calculate previous month's installs for each category
monthly_data["Previous_Installs"] = (
    monthly_data.groupby("Category_Display")["Installs"].shift(1)
)

# Calculate percentage growth
monthly_data["Growth_Percent"] = (
    (monthly_data["Installs"] - monthly_data["Previous_Installs"])
    / monthly_data["Previous_Installs"] * 100
)

# Identify growth greater than 20%
monthly_data["Significant_Growth"] = (
    monthly_data["Growth_Percent"] > 20
)

print("Growth calculation completed!")
display(
    monthly_data[
        ["Month", "Category_Display", "Installs",
         "Previous_Installs", "Growth_Percent",
         "Significant_Growth"]
    ].head(15)
)
```

```python

import plotly.graph_objects as go

fig = go.Figure()

# Add a separate line for each category
for category in monthly_data["Category_Display"].unique():
    cat_data = (
        monthly_data[
            monthly_data["Category_Display"] == category
        ].sort_values("Month").copy()
    )

    fig.add_trace(
        go.Scatter(
            x=cat_data["Month"],
            y=cat_data["Installs"],
            mode="lines+markers",
            name=category,
            connectgaps=False,
            hovertemplate=(
                "Month: %{x|%b %Y}<br>"
                "Category: " + category + "<br>"
                "Installs: %{y:,.0f}<extra></extra>"
            )
        )
    )

    # Shade the area under the curve where growth exceeds 20%
    growth_data = cat_data[
        cat_data["Significant_Growth"] == True
    ]

    if not growth_data.empty:
        fig.add_trace(
            go.Scatter(
                x=growth_data["Month"],
                y=growth_data["Installs"],
                mode="markers",
                marker=dict(color="red", size=10, symbol="diamond"),
                name=f"{category} — Growth >20%",
                showlegend=True,
                hovertemplate=(
                    "Month: %{x|%b %Y}<br>"
                    "Installs: %{y:,.0f}<br>"
                    "Growth >20%<extra></extra>"
                )
            )
        )

fig.update_layout(
    title="Monthly App Installs Trend by Category",
    xaxis_title="Month",
    yaxis_title="Total Installs",
    yaxis_type="log",
    template="plotly_white",
    hovermode="x unified",
    legend_title="App Category",
    height=650
)

fig.show()
```

```python

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
import re
from datetime import datetime
from zoneinfo import ZoneInfo

url = "https://raw.githubusercontent.com/vaish-vkb/Google-Play-Store-Analytics/main/Play%20Store%20Data.csv"

df4 = pd.read_csv(url)

print("Task 4 dataset loaded successfully!")
print("Rows and columns:", df4.shape)
print(df4.columns.tolist())
```

```python

# Convert required columns to numeric values
df4["Rating"] = pd.to_numeric(df4["Rating"], errors="coerce")
df4["Reviews"] = pd.to_numeric(df4["Reviews"], errors="coerce")

# Clean and convert Installs
df4["Installs"] = (
    df4["Installs"].astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)
df4["Installs"] = pd.to_numeric(df4["Installs"], errors="coerce")

# Convert app size to MB
def convert_size_to_mb(value):
    value = str(value).strip().upper()

    if value.endswith("M"):
        return pd.to_numeric(value[:-1], errors="coerce")
    elif value.endswith("K"):
        return pd.to_numeric(value[:-1], errors="coerce") / 1024
    else:
        return np.nan

df4["Size_MB"] = df4["Size"].apply(convert_size_to_mb)

# Convert Last Updated to datetime
df4["Last Updated"] = pd.to_datetime(
    df4["Last Updated"], errors="coerce"
)

# Apply Task 4 filters
filtered4 = df4[
    (df4["Rating"] >= 4.2) &
    (~df4["App"].astype(str).str.contains(r"\d", regex=True)) &
    (df4["Category"].str.startswith(("T", "P"))) &
    (df4["Reviews"] > 1000) &
    (df4["Size_MB"].between(20, 80))
].copy()

# Remove rows missing required values
filtered4 = filtered4.dropna(
    subset=["Installs", "Last Updated", "Category"]
)

print("Rows after filtering:", len(filtered4))
print("\nCategories included:")
print(sorted(filtered4["Category"].unique()))
print("\nSize range in MB:")
print(filtered4["Size_MB"].min(), "to", filtered4["Size_MB"].max())
```

```python

# Translate category names for the chart legend
category_translation4 = {
    "TRAVEL_AND_LOCAL": "Voyage et local",
    "PRODUCTIVITY": "Productividad",
    "PHOTOGRAPHY": "写真"
}

filtered4["Category_Display"] = (
    filtered4["Category"]
    .map(category_translation4)
    .fillna(filtered4["Category"])
)

# Create month from Last Updated
filtered4["Month"] = (
    filtered4["Last Updated"]
    .dt.to_period("M")
    .dt.to_timestamp()
)

# Aggregate installs by month and category
monthly4 = (
    filtered4.groupby(
        ["Month", "Category_Display"],
        as_index=False
    )["Installs"]
    .sum()
    .sort_values("Month")
)

print("Task 4 monthly data created!")
print("Rows:", len(monthly4))
print("\nTranslated categories:")
print(sorted(monthly4["Category_Display"].unique()))
display(monthly4.head(10))
```

```python

# Create a complete monthly timeline
all_months = pd.date_range(
    monthly4["Month"].min(),
    monthly4["Month"].max(),
    freq="MS"
)

# Pivot data for the stacked area chart
area_data = monthly4.pivot(
    index="Month",
    columns="Category_Display",
    values="Installs"
).reindex(all_months).fillna(0)

area_data.index.name = "Month"

# Identify months with >25% growth in any category
growth_months = set(
    monthly4.loc[
        monthly4["Significant_Growth"],
        "Month"
    ]
)

# Create stacked area chart
fig4 = go.Figure()

fig4_months = area_data.index

for category in area_data.columns:
    fig4.add_trace(
        go.Scatter(
            x=fig4_months,
            y=area_data[category],
            mode="lines",
            name=category,
            stackgroup="installs",
            hovertemplate=(
                "Month: %{x|%b %Y}<br>"
                + category
                + ": %{y:,.0f}<extra></extra>"
            )
        )
    )

# Highlight months with significant growth
for month in growth_months:
    if month in area_data.index:
        fig4.add_vrect(
            x0=month,
            x1=month + pd.offsets.MonthBegin(1),
            fillcolor="red",
            opacity=0.16,
            line_width=0,
            layer="above"
        )

fig4.update_layout(
    title="Monthly Installs by App Category — Task 4",
    xaxis_title="Month",
    yaxis_title="Installs",
    hovermode="x unified",
    template="plotly_white",
    height=650,
    legend_title="App Category"
)


from datetime import datetime
from zoneinfo import ZoneInfo

now_ist = datetime.now(ZoneInfo("Asia/Kolkata"))
current_minutes = now_ist.hour * 60 + now_ist.minute

start_minutes = 16 * 60  # 4 PM
end_minutes = 18 * 60    # 6 PM

if start_minutes <= current_minutes < end_minutes:
    print("Task 4 chart is available now (4 PM - 6 PM IST).")
    fig4.show()
else:
    print("Task 4 chart is hidden. Available daily from 4 PM to 6 PM IST.")

```

```python

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from datetime import datetime
from zoneinfo import ZoneInfo

url = "https://raw.githubusercontent.com/vaish-vkb/Google-Play-Store-Analytics/main/Play%20Store%20Data.csv"

df5 = pd.read_csv(url)

print("Task 5 dataset loaded successfully!")
print("Rows and columns:", df5.shape)
print(df5.columns.tolist())
```

```python

# Convert rating and reviews to numeric
df5["Rating"] = pd.to_numeric(df5["Rating"], errors="coerce")
df5["Reviews"] = pd.to_numeric(df5["Reviews"], errors="coerce")

# Clean and convert installs to numeric
df5["Installs"] = (
    df5["Installs"].astype(str)
    .str.replace(",", "", regex=False)
    .str.replace("+", "", regex=False)
)
df5["Installs"] = pd.to_numeric(df5["Installs"], errors="coerce")

# Convert app size to MB
def size_to_mb(value):
    value = str(value).strip().upper()

    if value.endswith("M"):
        return pd.to_numeric(value[:-1], errors="coerce")
    elif value.endswith("K"):
        return pd.to_numeric(value[:-1], errors="coerce") / 1024
    return np.nan

df5["Size_MB"] = df5["Size"].apply(size_to_mb)

# Convert Last Updated to datetime
df5["Last Updated"] = pd.to_datetime(
    df5["Last Updated"], errors="coerce"
)

# Apply rating, size, and January filters
filtered5 = df5[
    (df5["Rating"] >= 4.0) &
    (df5["Size_MB"] >= 10) &
    (df5["Last Updated"].dt.month == 1)
].copy()

filtered5 = filtered5.dropna(
    subset=["Category", "Installs", "Reviews", "Rating"]
)

print("Rows after filtering:", len(filtered5))
print("\nDate range:")
print(filtered5["Last Updated"].min(), "to",
      filtered5["Last Updated"].max())
print("\nCategories remaining:", filtered5["Category"].nunique())
```

```python

# Group apps by category
category_summary5 = (
    filtered5.groupby("Category", as_index=False)
    .agg(
        Average_Rating=("Rating", "mean"),
        Total_Reviews=("Reviews", "sum"),
        Total_Installs=("Installs", "sum")
    )
)

# Sort by total installs and select top 10 categories
top10_categories5 = (
    category_summary5
    .sort_values("Total_Installs", ascending=False)
    .head(10)
    .copy()
)

print("Top 10 categories by total installs:")
display(top10_categories5)
```

```python

# Prepare data for grouped bar chart
chart_data5 = top10_categories5.melt(
    id_vars="Category",
    value_vars=["Average_Rating", "Total_Reviews"],
    var_name="Metric",
    value_name="Value"
)

# Create grouped bar chart
fig5 = px.bar(
    chart_data5,
    x="Category",
    y="Value",
    color="Metric",
    barmode="group",
    title="Task 5: Average Rating vs Total Reviews — Top 10 Categories",
    labels={
        "Category": "App Category",
        "Value": "Value",
        "Metric": "Metric"
    },
    hover_data=["Value"]
)

fig5.update_layout(
    xaxis_tickangle=-45,
    height=650,
    legend_title="Comparison",
    template="plotly_white"
)

# Display chart only from 3 PM to 5 PM IST
from datetime import datetime
from zoneinfo import ZoneInfo

now_ist = datetime.now(ZoneInfo("Asia/Kolkata"))
current_minutes = now_ist.hour * 60 + now_ist.minute

if 15 * 60 <= current_minutes < 17 * 60:
    print("Task 5 chart is available now (3 PM - 5 PM IST).")
    fig5.show()
else:
    print("Task 5 chart is hidden. Available daily from 3 PM to 5 PM IST.")
```
